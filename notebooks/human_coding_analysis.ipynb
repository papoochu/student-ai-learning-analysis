```python

# Step 1: Define educational assistance categories

import pandas as pd

codebook = [
    {
        "Label": "Concept Explanation",
        "Definition": "Requests an explanation of a concept, principle, or idea.",
        "Example": "What is cross-validation?"
    },
    {
        "Label": "Debugging",
        "Definition": "Requests help identifying or fixing an error or malfunction.",
        "Example": "Why does my code produce a KeyError?"
    },
    {
        "Label": "Code Generation",
        "Definition": "Requests new code or completion of an implementation.",
        "Example": "Write a Python function that implements BFS."
    },
    {
        "Label": "Code Explanation / Review",
        "Definition": "Requests explanation, evaluation, or feedback on existing code.",
        "Example": "Can you explain what this function does?"
    },
    {
        "Label": "Problem Solving",
        "Definition": "Requests help solving a technical, mathematical, or analytical problem.",
        "Example": "Calculate the probability of this sequence."
    },
    {
        "Label": "Data Processing",
        "Definition": "Requests data cleaning, transformation, querying, or manipulation.",
        "Example": "Convert this SQL query into Pandas."
    },
    {
        "Label": "Assignment Guidance",
        "Definition": "Requests clarification of assignment instructions or requirements.",
        "Example": "What is part three asking me to do?"
    },
    {
        "Label": "Other / Unclear",
        "Definition": "The assistance type cannot be determined or does not fit.",
        "Example": "Do it again."
    }
]

codebook_df = pd.DataFrame(codebook)

display(codebook_df)

print("Number of categories:", len(codebook_df))
```

```python

# Add annotation rules to our existing codebook

coding_rules = [
    {
        "Rule": "Primary assistance type",
        "Guideline": (
            "Classify the action the student wants AI to perform, "
            "not merely the academic subject being discussed."
        )
    },
    {
        "Rule": "Overlapping categories",
        "Guideline": (
            "If multiple categories apply, select the student's "
            "main requested action as the primary label."
        )
    },
    {
        "Rule": "Debugging priority",
        "Guideline": (
            "If the student asks AI to diagnose or fix something "
            "that is not working, use Debugging, even if the "
            "request involves code or data."
        )
    },
    {
        "Rule": "Code generation vs. data processing",
        "Guideline": (
            "Use Code Generation for general code-writing requests. "
            "Use Data Processing when the primary goal is cleaning, "
            "converting, querying, or transforming data, even if "
            "the answer requires writing code."
        )
    },
    {
        "Rule": "Unclear follow-up requests",
        "Guideline": (
            "Check surrounding conversation context when available. "
            "If the intended assistance remains unclear, "
            "use Other / Unclear and flag the prompt."
        )
    }
]

coding_rules_df = pd.DataFrame(coding_rules)

display(coding_rules_df)
```

```python

# Update the Problem Solving category

problem_solving_definition = (
    "Requests a finished answer or completion of a question "
    "or task, rather than primarily an explanation. Includes "
    "mathematics, logical reasoning, logistical questions, "
    "copy-paste-ready answers, writing tasks, and other "
    "requests for completed solutions."
)

codebook_df.loc[
    codebook_df["Label"] == "Problem Solving",
    "Definition"
] = problem_solving_definition

codebook_df.loc[
    codebook_df["Label"] == "Problem Solving",
    "Example"
] = "Write a complete answer to question 5."

display(codebook_df)
```

```python

# Add rules for the revised Problem Solving category

new_rules = pd.DataFrame([
    {
        "Rule": "Problem Solving: finished answers",
        "Guideline": (
            "Use Problem Solving when the student primarily "
            "requests a completed answer or task, including "
            "math, logic, logistics, copy-paste-ready responses, "
            "or writing."
        )
    },
    {
        "Rule": "Problem Solving vs. Concept Explanation",
        "Guideline": (
            "Use Problem Solving when the student wants the "
            "answer. Use Concept Explanation when the student "
            "primarily wants to understand how or why "
            "something works."
        )
    },
    {
        "Rule": "Problem Solving vs. Code Generation",
        "Guideline": (
            "Use Code Generation when the requested finished "
            "product is programming code. Use Problem Solving "
            "for completed answers or tasks that are not "
            "primarily code generation."
        )
    },
    {
        "Rule": "Problem Solving vs. Data Processing",
        "Guideline": (
            "Use Data Processing for requests to clean, query, "
            "convert, transform, or format data, even when "
            "the student wants a finished result."
        )
    }
])

# Replace any previous rules with the same names
coding_rules_df = coding_rules_df[
    ~coding_rules_df["Rule"].isin(new_rules["Rule"])
].copy()

coding_rules_df = pd.concat(
    [coding_rules_df, new_rules],
    ignore_index=True
)

display(coding_rules_df)
```

```python

# Update Code Generation and Data Processing definitions

codebook_df.loc[
    codebook_df["Label"] == "Code Generation",
    "Definition"
] = (
    "Requests AI to write, complete, translate, or convert "
    "programming code. Includes translating SQL queries into "
    "Pandas/Python code, implementing functions, and generating "
    "new code from specifications."
)

codebook_df.loc[
    codebook_df["Label"] == "Code Generation",
    "Example"
] = "Convert this SQL query into Pandas code."

codebook_df.loc[
    codebook_df["Label"] == "Data Processing",
    "Definition"
] = (
    "Requests cleaning, formatting, transformation, analysis, "
    "or manipulation of data itself. Use when the requested "
    "output is processed data, a table, or a data result "
    "rather than newly written programming code."
)

codebook_df.loc[
    codebook_df["Label"] == "Data Processing",
    "Example"
] = "Remove duplicate rows from this dataset."

display(
    codebook_df[
        codebook_df["Label"].isin(
            ["Code Generation", "Data Processing"]
        )
    ]
)
```

```python

# Replace conflicting coding rules

coding_rules_df = coding_rules_df[
    ~coding_rules_df["Rule"].isin([
        "Code generation vs. data processing",
        "Problem Solving vs. Data Processing",
        "SQL to Pandas conversion"
    ])
].copy()

new_rules = pd.DataFrame([
    {
        "Rule": "Code generation vs. data processing",
        "Guideline": (
            "Use Code Generation when the requested output is "
            "new, completed, or translated code, including "
            "SQL-to-Pandas conversions. Use Data Processing "
            "when the requested output is cleaned, formatted, "
            "transformed, or analyzed data itself."
        )
    },
    {
        "Rule": "Problem Solving vs. Data Processing",
        "Guideline": (
            "Use Data Processing when the main task is to "
            "manipulate or present data. Use Problem Solving "
            "for completed answers to other questions or tasks, "
            "including math, logic, logistics, and writing."
        )
    },
    {
        "Rule": "SQL to Pandas conversion",
        "Guideline": (
            "Always classify requests to translate SQL queries "
            "into Pandas or Python code as Code Generation."
        )
    }
])

coding_rules_df = pd.concat(
    [coding_rules_df, new_rules],
    ignore_index=True
)

display(coding_rules_df.tail(3))
```

```python

# Step 2: Load StudyChat and sample 200 prompts for human coding

import pandas as pd
import numpy as np

df = pd.read_json(
    "hf://datasets/wmcnicho/StudyChat/data.jsonl",
    lines=True
)

# Match the cleaning rules used in our previous notebooks
modeling_corpus = df.dropna(subset=["prompt"]).copy()

modeling_corpus = modeling_corpus[
    modeling_corpus["prompt"].astype(str).str.strip() != ""
].copy()

word_counts = modeling_corpus["prompt"].astype(str).str.split().str.len()
modeling_corpus = modeling_corpus[
    word_counts <= 1000
].copy()

modeling_corpus = modeling_corpus.reset_index(drop=True)

# Sample one prompt per student, then select 200 students
one_per_student = (
    modeling_corpus
    .groupby("userId", group_keys=False)
    .sample(n=1, random_state=42)
)

annotation_sample = (
    one_per_student
    .sample(n=200, random_state=42)
    .reset_index(drop=True)
)

# Create human-labeling columns
annotation_sample["sample_id"] = [
    f"P{i:03d}" for i in range(1, len(annotation_sample) + 1)
]

annotation_sample["human_label"] = ""
annotation_sample["ambiguous"] = ""
annotation_sample["notes"] = ""

print("Total cleaned prompts:", len(modeling_corpus))
print("Annotation sample:", len(annotation_sample))
print("Students represented:", annotation_sample["userId"].nunique())

display(
    annotation_sample[
        ["sample_id", "prompt", "human_label", "ambiguous", "notes"]
    ].head(10)
)
```

```python

# Step 3b: Save and download annotation files

from google.colab import files

# File 1: The 200 prompts to label
annotation_sample[
    [
        "sample_id",
        "userId",
        "chatId",
        "prompt",
        "human_label",
        "ambiguous",
        "notes"
    ]
].to_csv(
    "studychat_human_annotation_200.csv",
    index=False
)

# File 2: Our eight assistance categories
codebook_df.to_csv(
    "studychat_assistance_codebook.csv",
    index=False
)

# File 3: Our annotation rules
coding_rules_df.to_csv(
    "studychat_coding_rules.csv",
    index=False
)

# Download all three files
files.download("studychat_human_annotation_200.csv")
files.download("studychat_assistance_codebook.csv")
files.download("studychat_coding_rules.csv")

print("All three annotation files saved!")
```
